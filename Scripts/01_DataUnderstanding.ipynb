## Summary of Data Understanding
The dataset contains user activity logs from a music streaming service. Each record represents an event performed by a user, such as playing a song, logging in, or upgrading to a paid account. The dataset includes various attributes related to user demographics, device information, and song details.

### Key Findings:
- The dataset contains logs of user activities with timestamps, user IDs, session IDs, and event types.
- The data has time range from 2018-10-01 to 2018-12-03.
- The data are from 226 users, with a total of 277.4k records.
- 52 users have "Cancellation Confirmation" events, indicating a churn rate of approximately 23%.
- 7 / 52 users churned before 2018-10-08, which may not have enough data for training.

### Questions Required Clarification:
- Definition of "churn"? Currently defined as users who have page event "Cancellation Confirmation", with status 200 as successful cancellation.
- Definition of page events, e.g. Upgrade / Downgrade
- How will this model be used in production? Real-time or batch processing? This will affect formation of cohort for training and evaluation.
- How is the data collected? Any sampling bias?
